```python
# Get all unique tasks
all_tasks = set()

for entry in df['What tasks do you use AI for? '].dropna():
    tasks = [task.strip() for task in str(entry).split(';') if task.strip()]
    all_tasks.update(tasks)

# Create new columns
for task in all_tasks:
    df[task] = df['What tasks do you use AI for? '].apply(
        lambda x: 1 if pd.notna(x) and task in [t.strip() for t in str(x).split(';')] else 0
    )

# View updated table
print(df.head())
```
