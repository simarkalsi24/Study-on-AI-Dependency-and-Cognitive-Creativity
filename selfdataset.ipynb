```python
cols = [
    'Were you aware that AI could generate write-ups?',
    'How mentally tiring was the task?',
    'How confident are you in your originality?',
    'Were you tempted to use AI while writing?'
]

for col in cols:
    df1[col] = df1[col].fillna(df1[col].mode()[0])
```

```python
df1.isna().sum()
```

```python
df1['If yes, how often?'] = df1['If yes, how often?'].fillna('Not Applicable')
df1['What tasks do you use AI for? '] = df1['What tasks do you use AI for? '].fillna('Not Applicable')
```

```python
import matplotlib.pyplot as plt

df1['Age '].hist()
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

```python
import seaborn as sns

cross_tab = pd.crosstab(
    df1['On a scale of 1–5, how dependent do you feel on AI?'],
    df1['Do you feel AI has reduced your own creativity?']
)

sns.heatmap(cross_tab, annot=True, fmt='d')
plt.title("AI Dependency vs Creativity")
plt.show()
```

```python
pd.crosstab(
    df1['Were you tempted to use AI while writing?'],
    df1['Do you regularly use AI tools (e.g., ChatGPT, Grammarly)']
).plot(kind='bar')

plt.title("Temptation vs AI Usage")
plt.show()
```
